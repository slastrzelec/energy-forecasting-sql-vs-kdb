```python
import boto3

s3 = boto3.client('s3')

response = s3.list_buckets()

print(response)

for bucket in response['Buckets']:
    print(f'Bucket Name: {bucket["Name"]}, Creation Date: {bucket["CreationDate"]}')    


```

```python
bucket_name = 'londonss'
s3 = boto3.client('s3')
response = s3.list_objects_v2(Bucket=bucket_name)

print(f"Pliki w buckecie '{bucket_name}':")
print("="*50)

if 'Contents' in response:
    for obj in response['Contents']:
        key = obj['Key']
        size = obj['Size']
        date = obj['LastModified']
        print(f"  {key} ({size} bytes) - {date}")
else:
    print("Bucket jest pusty")
```

```python
import boto3
import pandas as pd
import io

s3 = boto3.client('s3')

# Pobierz konkretny plik
response = s3.get_object(
    Bucket='londonss', 
    Key='athena-results/21abbd59-f2ca-4a91-a92b-4596a60d79b2.csv'
)

# Wczytaj do DataFrame
df = pd.read_csv(io.BytesIO(response['Body'].read()))

# Wyświetl
print(df)
```
