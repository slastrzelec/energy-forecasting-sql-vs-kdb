```python
import boto3
from dotenv import load_dotenv
import os

# 1. Załaduj zmienne z pliku .env
load_dotenv()  # domyślnie szuka pliku .env w katalogu projektu

# 2. Pobierz wartości
aws_access_key_id = os.getenv('AWS_ACCESS_KEY_ID')
aws_secret_access_key = os.getenv('AWS_SECRET_ACCESS_KEY')
region_name = os.getenv('AWS_DEFAULT_REGION')

# 3. Stwórz klienta S3
s3 = boto3.client(
    's3',
    aws_access_key_id=aws_access_key_id,
    aws_secret_access_key=aws_secret_access_key,
    region_name=region_name
)

# 4. Sprawdź istniejące bucket-y
response = s3.list_buckets()
for bucket in response['Buckets']:
    print(bucket['Name'])

print(response)
```

```python
import boto3
from dotenv import load_dotenv
import os

load_dotenv()

s3 = boto3.client(
    's3',
    aws_access_key_id=os.getenv('AWS_ACCESS_KEY_ID'),
    aws_secret_access_key=os.getenv('AWS_SECRET_ACCESS_KEY'),
    region_name=os.getenv('AWS_DEFAULT_REGION')
)

bucket_name = 'demografia-pl-2025'

# 1. Usuń wszystkie obiekty w bucket
response = s3.list_objects_v2(Bucket=bucket_name)
if 'Contents' in response:
    for obj in response['Contents']:
        s3.delete_object(Bucket=bucket_name, Key=obj['Key'])
    print("Usunięto wszystkie pliki w bucket.")
else:
    print("Bucket jest już pusty.")

# 2. Usuń sam bucket
s3.delete_bucket(Bucket=bucket_name)
print(f"Bucket '{bucket_name}' został usunięty.")
```

```python
import boto3
from boto3.s3.transfer import TransferConfig
from dotenv import load_dotenv
import os

# 1️⃣ Załaduj zmienne AWS z pliku .env
load_dotenv()

bucket_name = 'londonss'
local_file = r"C:\Users\slast\PYTHON\0_projekty do portfolio\08_AWS_proj\data\raw\LCL_2013.csv"
s3_key = 'raw/LCL_2013.csv'  # folder w bucket + nazwa pliku

# 2️⃣ Konfiguracja uploadu wieloczęściowego
config = TransferConfig(
    multipart_threshold=5*1024*1024,    # pliki >5MB dzielone na części
    multipart_chunksize=25*1024*1024,   # każda część 25 MB
    max_concurrency=4,                   # równoległe wysyłanie 4 części
    use_threads=True
)

# 3️⃣ Utworzenie klienta S3
s3 = boto3.client(
    's3',
    aws_access_key_id=os.getenv('AWS_ACCESS_KEY_ID'),
    aws_secret_access_key=os.getenv('AWS_SECRET_ACCESS_KEY'),
    region_name=os.getenv('AWS_DEFAULT_REGION')
)

# 4️⃣ Wysyłka pliku
try:
    s3.upload_file(local_file, bucket_name, s3_key, Config=config)
    print(f"Plik {local_file} został wysłany do {bucket_name}/{s3_key} z użyciem multipart upload")
except Exception as e:
    print("Wystąpił błąd podczas uploadu:", e)
```
